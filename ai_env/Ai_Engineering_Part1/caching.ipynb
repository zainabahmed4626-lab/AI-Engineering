### Step 1: Install Open API and Libraries

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display
import gradio as gr

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API key is missing")
```

### Step 2: Install Lite LLM


```python
from litellm import completion

messages = [{"role": "user", "content": "Where is Surfers Paradise?"}]

response = completion(
    model="openai/gpt-5-nano",
    messages=messages
)

reply = response.choices[0].message.content
display(Markdown(reply))
```

### Step 3: :LLM Call Info

```python
import json

print(json.dumps(response.model_dump(), indent=2))
```

```python
print(f"prompt tokens: {response.usage.prompt_tokens}")
print(f"completion tokens: {response.usage.completion_tokens}")
print(f"total tokens: {response.usage.total_tokens}")
```

```python
import json

print(json.dumps(response._hidden_params, indent=2))
```

```python
print(f"cost: ${response._hidden_params['response_cost']:.6f}")
```
