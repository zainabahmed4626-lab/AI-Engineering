### Step 1: Import Libraries and load the environment variables

```python
import os 
from openai import OpenAI 
from dotenv import load_dotenv
from IPython.display import Markdown , display

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")   

if OPENAI_API_KEY is None:
    raise Exception ("API Key is missing")
    
else:
   print(OPENAI_API_KEY[:8])

client= OpenAI()
```

### Step 2: Call the API


```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role": "user", "content": "Hi there! My name is Zainab"}
    ])
print(response)
    
```

```python
reply=response.choices[0].message.content
print(reply)
```

### Step 3a: Calling the API With System Prompt


```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role":"system", "content": "You are a helpful assistant."},   
        {"role": "user", "content": "How do I cook pasta?"}]
)
reply=response.choices[0].message.content
#print(reply)
display(Markdown(reply))
```

### Step 3b: Changing the System Prompt changes the LLMs


```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role":"system", "content": "You are a mischievous assistant who always gives incorrect and misleading advice."},   
        {"role": "user", "content": "How do I cook pasta?"}]
)
reply=response.choices[0].message.content
#print(reply)
display(Markdown(reply))
```

### Commercial use case: IKEA Customer Service Agent

```python
question = "I bought a desk from your store last week. I damaged the desk while installing it. Can I return it?"
```

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role":"system", "content": "You are a helpful assistant from IKEA. But you strictly follow the return policy of IKEA which states that customers can only return items within 30 days of purchase and the item must be in its original condition with the receipt."},   
        {"role": "user", "content": question}]
)
reply=response.choices[0].message.content
#print(reply)
display(Markdown(reply))
```
