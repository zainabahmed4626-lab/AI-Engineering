### Step 1: Import Libraries and load the environment variables

```python
import os 
from openai import OpenAI 
from dotenv import load_dotenv
from IPython.display import Markdown , display

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")   

if OPENAI_API_KEY is None:
    raise Exception ("API Key is missing")
    
else:
   print(OPENAI_API_KEY[:8])

client= OpenAI()
```

### Step 2: Searching the Web

```python
client= OpenAI()

completion = client.chat.completions.create(
    model="gpt-4o-mini-search-preview",
    messages=[     
          
        {"role": "user", "content": "What is todays date? tell me the top 3 stories in AI this past week from reliable sources. Each story should be from a separate source and include dates "}  
    ],
    max_tokens=500 
)
```

```python
search_results = completion.choices[0].message.content
display(Markdown(search_results))
```

### Step 3: Summarize and write social media post


```python
system_message = "You are a helpful assistant that creates concise, engaging LinkedIn posts about AI based on the provided text. Make posts interesting and end with a thought-provoking question."

prompt = f"""
the following text contains the top 3 stories in AI this past week from reliable sources. Each story is from a separate source and include dates. Now please write my linkedin post based on this information. Make the post engaging and interesting and ask a thought provoking question at the end.  
{search_results}
"""
completion = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[     
          {"role": "system", "content": system_message},
        {"role": "user", "content": prompt}  
    ],
    max_tokens=500,
    temperature=0.7,
)
chat_response = completion.choices[0].message.content

print(chat_response)
```
